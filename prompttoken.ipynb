```python
import tiktoken

tokenizer=tiktoken.encoding_for_model("gpt-4o")
```

```python
print(tokenizer.name)

tokenizer=tiktoken.get_encoding("o200k_base")
```

```python
prompt="Vous êtes un expert en marketing digital. C'est quoi un digital marketer ? "

tokens=tokenizer.encode(prompt)

print(tokens)

print("                  ")

print(len(tokens))

print("                  ")


for token in tokens:
    t=tokenizer.decode_single_token_bytes(token).decode("utf-8")
    print(t,end='|')
```

# Utilisation des llm

```python
#with gemini cles 

from dotenv.ipython import load_dotenv
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from IPython.display import display, Markdown

# Charger  clé API de  .env (file li dart fih cle direct using aistudio)
load_dotenv(override=True)

# Initialiser model
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0
)

# send question l modèle
resp = llm.invoke([
    ("system", "Vous êtes un expert en marketing digital."),
    ("human", "C'est quoi un digital marketer ?")
])

# Afficher la réponse en Markdown
display(Markdown(resp.text))
```
